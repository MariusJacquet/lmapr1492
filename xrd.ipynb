## Partie 3 : Pics du diffractogramme pour un rayonnement incident Cu Ka de la molécules $Cs_2O$

```python
from pymatgen.ext.matproj import MPRester
from pymatgen.analysis.diffraction.xrd import XRDCalculator
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
```

```python
mp_key = "ClP8wE7rgC4VunNMMiut1b2BYMfgW6Lu"
mp_id = "mp-7988"
```

```python
with MPRester(mp_key) as m:
    prim_struc = m.get_structure_by_material_id(mp_id)
conv_struc = SpacegroupAnalyzer(prim_struc).get_conventional_standard_structure()
```

```python
xrd = XRDCalculator(wavelength = "CuKa")
pattern = xrd.get_pattern(conv_struc)
```

## Calcul des 3 premiers pics du diffractogramme

```python
i = 0
pic = ["premier", "deuxième", "troisième"]
print("----------------------------------------------------------------------------\n")
while i < 3:
    print("Pic du diffractogramme n°",i+1,"\n",
         "Angle 2Θ entre les 2 rayons :", pattern.x[i],"\n",
         "Intensité du", pic[i],"pic :", pattern.y[i],"\n",
         "Indices hkl associés :", pattern.hkls[i],"\n",
         "Distance associée :", pattern.d_hkls[i],"\n",
         "\n ----------------------------------------------------------------------------\n")
    i+=1
```
